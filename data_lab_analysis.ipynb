# Hands-On Data Lab Implementation
## Sales Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('sales_data.csv')
df.head()
```

```python
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
```

```python
df['Region']=df['Region'].fillna('Unknown')
df[['Units_Sold','Unit_Price','Revenue']].describe()
```

```python
summary=df.groupby('Product').agg(Total_Units=('Units_Sold','sum'),Total_Revenue=('Revenue','sum')).sort_values('Total_Revenue',ascending=False)
summary
```

```python
df[['Units_Sold','Unit_Price','Revenue']].corr().round(2)
```

```python
summary['Total_Revenue'].plot(kind='bar',figsize=(8,5)); plt.title('Total Revenue by Product'); plt.tight_layout(); plt.show()
```
